# Аналитический отчёт по данным MovieLens

## Цель исследования

Цель отчёта — исследовать пользовательские оценки фильмов и определить:

- как распределяются оценки;
- какие пользователи наиболее активны;
- какие фильмы наиболее популярны;
- какие фильмы получили самые высокие и низкие средние оценки;
- какие жанры чаще встречаются среди лучших и худших фильмов;
- как средний рейтинг фильмов меняется в зависимости от десятилетия выпуска.

Для анализа используются данные из файлов `ratings.csv` и `movies.csv`.
Все вычисления выполняются методами модуля `movielens_analysis.py`.

```python
from movielens_analysis import Movies, Ratings

%timeit -n 1 -r 1 from movielens_analysis import Movies, Ratings
```

```python
movies = Movies()
ratings = Ratings()

%timeit -n 1 -r 1 Movies(), Ratings()
```

```python
ratings_list = ratings.get_ratings()
movies_list = movies.get_movies()

%timeit -n 1 -r 1 ratings.get_ratings()
%timeit -n 1 -r 1 movies.get_movies()
```

## Загрузка данных

Файл `ratings.csv` содержит идентификатор пользователя, идентификатор
фильма и выставленную оценку.

Файл `movies.csv` содержит идентификатор фильма, его название и список
жанров.

На следующем этапе рассмотрим распределение пользовательских оценок.

```python
rating_counts = ratings.count_ratings(ratings_list)

%timeit ratings.count_ratings(ratings_list)

rating_counts
```

## Распределение пользовательских оценок

Наиболее часто пользователи ставили оценку 4.0. Реже всего
встречается оценка 0.5.

Распределение показывает, что пользователи чаще ставят
положительные и средние оценки. Это может означать, что ...

```python
active_users = ratings.sort_user(ratings_list)

%timeit ratings.sort_user(ratings_list)

active_users
```

## Самые активные пользователи

В результате анализа были определены десять пользователей, которые
оставили больше всего оценок.

Самым активным оказался пользователь с идентификатором 414,
который оценил 2698 фильмов.

Результат показывает, что небольшая группа пользователей может вносить
значительный вклад в формирование базы данных.

```python
sorted_films = movies.get_all_films_information(
    ratings_list,
    movies_list,
)

%timeit -n 1 -r 1 movies.get_all_films_information(ratings_list, movies_list)

sorted_films[:5]
```

## Формирование общей статистики фильмов

Информация из `ratings.csv` и `movies.csv` была объединена по
идентификатору фильма.

Для каждого фильма были рассчитаны:

- средний пользовательский рейтинг;
- количество полученных оценок;
- жанры;
- год выпуска.

Фильмы без пользовательских оценок были исключены из дальнейшего
анализа.

```python
most_reviewed_films = movies.get_most_rev_films(sorted_films)

%timeit -n 1 -r 1 movies.get_most_rev_films(sorted_films)

most_reviewed_films
```

## Самые популярные фильмы

Популярность фильма определяется количеством пользовательских оценок,
а не значением среднего рейтинга.

Наибольшее количество оценок получил фильм Forrest Gump — его оценили
329 пользователей. Его средний рейтинг составляет 4.16.

Можно заметить, что популярность фильма не обязательно означает
максимальный средний рейтинг.

## Лучшие и худшие фильмы

Фильм с одной оценкой `5.0` нельзя уверенно считать лучшим. Поэтому в
анализ включаются только фильмы, получившие больше 20 пользовательских
оценок.

Лучшие и худшие фильмы определяются по среднему рейтингу.

```python
top_100_films = movies.get_top_100_films(sorted_films)

%timeit -n 1 -r 1 movies.get_top_100_films(sorted_films)

top_100_films[:10]
```

### Лучшие фильмы

В список вошли фильмы с наибольшим средним рейтингом, получившие более
20 оценок.

Первое место занимает фильм Побег из Шоушенка со средним рейтингом 4.42.
Количество пользовательских оценок этого фильма — 317.

Использование ограничения по количеству оценок делает результат более
надёжным и исключает фильмы, высокий рейтинг которых основан на
единичных отзывах.

```python
worst_100_films = movies.get_worst_100_films(sorted_films)

%timeit -n 1 -r 1 movies.get_worst_100_films(sorted_films)

worst_100_films[:10]
```

### Худшие фильмы

Список сформирован из фильмов с наиболее низким средним рейтингом,
получивших больше 20 пользовательских оценок.

Самый низкий результат у фильма Анаконда. Его средняя оценка составляет
1.92, а количество полученных оценок - 27.

Порог по количеству оценок позволяет считать низкий рейтинг устойчивым,
а не результатом мнения одного пользователя.

```python
interesting_films = movies.get_interesting_films(sorted_films)

%timeit -n 1 -r 1 movies.get_interesting_films(sorted_films)

interesting_films
```

## Малоизвестные высокооценённые фильмы

В этом разделе представлены фильмы, которые получили от 5 до 19 оценок,
но имеют высокий средний рейтинг.

Такие фильмы можно назвать «скрытыми жемчужинами»: они получили
положительные отзывы, но пока не приобрели большой популярности.

Самый высокий средний рейтинг в этой категории у фильма Trial, The (Procès, Le) —
4.9 при 5 пользовательских оценках.

## Анализ жанров

Сначала определим, какие жанры чаще всего встречаются среди ста лучших
и ста худших фильмов.

Если фильм относится к нескольким жанрам, он учитывается один раз в
каждом из них.

```python
top_genres = movies.get_top_genres(top_100_films)

%timeit -n 1 -r 1movies.get_top_genres(top_100_films)

top_genres
```

```python
worst_genres = movies.get_worst_genres(worst_100_films)

%timeit -n 1 -r 1 movies.get_worst_genres(worst_100_films)

worst_genres
```

### Жанры лучших и худших фильмов

Наиболее распространённым жанром среди лучших фильмов оказалась
Драма — он встречается в 59 фильмах из ста.

Среди худших фильмов чаще всего встречается жанр Комедия —
он представлен в 51 фильмах.

Некоторые жанры могут одновременно находиться в обоих списках. Это не
обязательно означает, что жанр является хорошим или плохим: такой
результат может быть связан с его общей распространённостью в наборе
данных.

```python
average_genre_ratings = movies.get_av_genres_rating(sorted_films)

%timeit -n 1 -r 1 movies.get_av_genres_rating(sorted_films)

average_genre_ratings.most_common()
```

### Средний рейтинг жанров

Для каждого жанра было рассчитано среднее значение средних рейтингов
всех относящихся к нему фильмов.

Наиболее высокий результат получил жанр Документальное со средним рейтингом
3.78. Самый низкий результат у жанра Хоррор — 2.91.

При интерпретации следует учитывать, что каждый фильм имеет одинаковый
вес независимо от количества пользовательских оценок. Кроме того, один
фильм может одновременно относиться к нескольким жанрам.

## Анализ фильмов по десятилетиям

Фильмы были объединены в группы по десятилетию выпуска. Для каждого
десятилетия рассчитываются:

- среднее значение средних рейтингов фильмов;
- общее количество пользовательских оценок;
- количество оценённых фильмов.

Фильмы, в названии которых отсутствует год выпуска, не учитываются.

```python
year_analytics = movies.get_year_analystics(sorted_films)

%timeit -n 1 -r 1 movies.get_year_analystics(sorted_films)

year_analytics
```

### Результаты анализа десятилетий

Наиболее высокий средний рейтинг получили фильмы 40-х годов —
3.63.

Больше всего оценённых фильмов относится к 00-м годам, а наибольшее
количество пользовательских оценок получили фильмы 90-х годов.

Результаты позволяют сравнить не только качество фильмов разных
периодов, но и их представленность и популярность в MovieLens.

Следует учитывать, что средний рейтинг десятилетия рассчитывается как
среднее от средних рейтингов фильмов. Каждый фильм имеет одинаковый вес
независимо от количества пользовательских оценок.

```python
example_year = movies.extract_year("Toy Story (1995)")

%timeit -n 1 -r 1 movies.extract_year("Toy Story (1995)")

example_year
```

## Извлечение года выпуска

Год выпуска фильма извлекается из его названия. Например, для фильма
`Toy Story (1995)` метод возвращает значение `1995`.

Если название не заканчивается годом в круглых скобках, метод возвращает
`None`.

```python
sorted_ratings = ratings.get_sort_ratings(ratings_list)

%timeit -r 1 -n 1 ratings.get_sort_ratings(ratings_list)

sorted_ratings[:5]
```

## Сортировка пользовательских оценок

Пользовательские оценки были отсортированы по убыванию. Первые элементы
списка содержат оценки `5.0`.

Эта сортировка показывает максимальные отдельные оценки, но не
используется для определения лучших фильмов. Для сравнения фильмов
необходимо учитывать их средний рейтинг и количество полученных оценок.

Основные результаты:

1. Наиболее распространённой пользовательской оценкой является 4.0.

3. Самым популярным по количеству оценок стал фильм Форрест Гамп.
4. Наиболее высокий устойчивый средний рейтинг среди фильмов с более

Исследование показывает, что популярность и средний рейтинг — разные

оценок помогает получать более устойчивые результаты и уменьшает
влияние случайных единичных отзывов.
